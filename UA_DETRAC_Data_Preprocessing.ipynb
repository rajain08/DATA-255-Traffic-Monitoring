```python
#Import libraries
import random
import shutil
import zipfile
import xml.etree.ElementTree as ET
from pathlib import Path

import gdown
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import yaml
from PIL import Image, ImageDraw
from sklearn.model_selection import train_test_split

#Set random seed
random.seed(42)
np.random.seed(42)
```

2. Set Up Folders and Variables

```python
#Main project folders
project_dir = Path('/content/traffic-monitoring') if Path('/content').exists() else (Path.cwd() if Path.cwd().name == 'traffic-monitoring' else Path.cwd() / 'traffic-monitoring')
zip_dir = project_dir / 'data' / 'zip'
raw_dir = project_dir / 'data' / 'raw'
processed_dir = project_dir / 'data' / 'processed'
figure_dir = project_dir / 'outputs' / 'figures'

#Create folders
for folder in [zip_dir, raw_dir, processed_dir, figure_dir]:
    folder.mkdir(parents=True, exist_ok=True)

#Dataset settings
max_sequences = 56
frame_step = 5

#Vehicle classes used by UA-DETRAC
class_names = ['car', 'bus', 'van', 'others']
class_ids = {name: i for i, name in enumerate(class_names)}
```

3. Extract Dataset Archives


```python
#Extract all zip files from data/zip into data/raw if not already extracted
zip_files = sorted(zip_dir.glob('*.zip'))

for zip_file in zip_files:
    if 'Images' in zip_file.name:
        target_dir = raw_dir / 'DETRAC-Images'
        already_extracted = target_dir.exists() and any(target_dir.iterdir())
    else:
        already_extracted = any(raw_dir.rglob(f'{zip_file.stem}/*.xml')) or any(raw_dir.glob(f'{zip_file.stem}/*.xml'))

    if not already_extracted:
        print(f'Extracting: {zip_file.name} to {raw_dir}')
        with zipfile.ZipFile(zip_file, 'r') as zip_ref:
            zip_ref.extractall(raw_dir)
    else:
        print(f'Already extracted: {zip_file.name}')

print('Extraction step complete')
```

4. Match Image Sequences with XML Files


```python
#Find XML files in data/raw
xml_files = sorted(raw_dir.rglob('MVI_*.xml'))

#Find folders that contain sequence images in data/raw
image_folders = {}
for folder in (raw_dir / 'DETRAC-Images').glob('MVI_*'):
    if folder.is_dir() and list(folder.glob('img*.jpg')):
        image_folders[folder.name] = folder

#Match each sequence with its XML file
def get_camera_group(sequence_name):
    prefix = sequence_name[:-1]
    if prefix in ('MVI_3950', 'MVI_3951'):
        return 'MVI_3950'
    if prefix in ('MVI_3980', 'MVI_3981', 'MVI_3982'):
        return 'MVI_3980'
    if prefix in ('MVI_4073', 'MVI_4075'):
        return 'MVI_4075'
    if prefix in ('MVI_4106', 'MVI_4107'):
        return 'MVI_4106'
    return prefix

sequence_data = []

for xml_file in xml_files:
    sequence_name = xml_file.stem

    if sequence_name in image_folders:
        root = ET.parse(xml_file).getroot()
        seq_attr = root.find('sequence_attribute')
        weather = seq_attr.get('sence_weather', 'unknown') if seq_attr is not None else 'unknown'
        frames = root.findall('./frame')
        object_count = sum(len(frame.findall('./target_list/target')) for frame in frames)

        sequence_data.append({
            'sequence': sequence_name,
            'camera_group': get_camera_group(sequence_name),
            'image_folder': image_folders[sequence_name],
            'xml_file': xml_file,
            'weather': weather,
            'frames': len(frames),
            'objects': object_count
        })

sequence_df = pd.DataFrame(sequence_data)

if len(sequence_df) == 0:
    raise ValueError('No matching image and XML sequences were found')

print(f'Found {len(image_folders)} image sequences and {len(xml_files)} XML files.')
print('Matched sequences:', len(sequence_df))
display(sequence_df[['sequence', 'weather', 'frames', 'objects']].head())
```

5. Split by Video Sequence

The split is done by complete sequence instead of individual frame. This prevents nearly identical frames from appearing in both training and testing.

```python
#Group sequences by camera site so neighboring sequences stay in the same split
camera_groups = (
    sequence_df.groupby(['camera_group', 'weather'], as_index=False)
    .agg(seq_count=('sequence', 'count'), frames=('frames', 'sum'))
)
camera_groups['avg_frames'] = camera_groups['frames'] / camera_groups['seq_count']

#Stratify by weather (9 train, 2 val, 3 test per weather) and balance sequence lengths for a 70/15/15 frame split
per_weather = max_sequences // sequence_df['weather'].nunique() if max_sequences is not None else None
test_target = 3
val_target = 2
train_target = (per_weather - test_target - val_target) if per_weather is not None else 999

split_dict = {}
selected_cameras = []

for weather, weather_cameras in camera_groups.groupby('weather'):
    used_cameras = set()
    non_overpass = weather_cameras[~weather_cameras['camera_group'].str.startswith('MVI_200')]

    test_candidates = non_overpass.iloc[(non_overpass['avg_frames'] - 700).abs().argsort()]
    test_count = 0
    for cam in test_candidates.itertuples():
        if test_count + cam.seq_count <= test_target:
            split_dict[cam.camera_group] = 'test'
            test_count += cam.seq_count
            used_cameras.add(cam.camera_group)
            selected_cameras.append(cam.camera_group)

    val_candidates = non_overpass[~non_overpass['camera_group'].isin(used_cameras)]
    val_candidates = val_candidates.iloc[(val_candidates['avg_frames'] - 1500).abs().argsort()]
    val_count = 0
    for cam in val_candidates.itertuples():
        if val_count + cam.seq_count <= val_target:
            split_dict[cam.camera_group] = 'val'
            val_count += cam.seq_count
            used_cameras.add(cam.camera_group)
            selected_cameras.append(cam.camera_group)

    train_candidates = weather_cameras[~weather_cameras['camera_group'].isin(used_cameras)].sort_values(
        ['avg_frames', 'seq_count'], ascending=[False, False]
    )
    train_count = 0
    for cam in train_candidates.itertuples():
        if train_count + cam.seq_count <= train_target:
            split_dict[cam.camera_group] = 'train'
            train_count += cam.seq_count
            selected_cameras.append(cam.camera_group)

sequence_df = sequence_df[sequence_df['camera_group'].isin(selected_cameras)].copy().reset_index(drop=True)
sequence_df['split'] = sequence_df['camera_group'].map(split_dict)

display(sequence_df[['sequence', 'camera_group', 'weather', 'split']])
print(sequence_df['split'].value_counts())
display(pd.crosstab(sequence_df['weather'], sequence_df['split'], margins=True))
```

6. Convert Bounding Boxes to YOLO Format

```python
def get_vehicle_type(target):
    #Read the vehicle type from the XML target
    attributes = target.find('./attribute')

    if attributes is None:
        return 'others'

    vehicle_type = attributes.attrib.get('vehicle_type', 'others').lower()

    if vehicle_type == 'other':
        vehicle_type = 'others'

    if vehicle_type not in class_names:
        vehicle_type = 'others'

    return vehicle_type


def convert_box(box, image_width, image_height):
    #Read original pixel coordinates
    left = float(box.attrib['left'])
    top = float(box.attrib['top'])
    width = float(box.attrib['width'])
    height = float(box.attrib['height'])

    #Keep the box inside the image
    left = max(0, min(left, image_width))
    top = max(0, min(top, image_height))
    right = max(0, min(left + width, image_width))
    bottom = max(0, min(top + height, image_height))

    width = right - left
    height = bottom - top

    if width <= 1 or height <= 1:
        return None

    #Convert to normalized YOLO coordinates
    x_center = (left + width / 2) / image_width
    y_center = (top + height / 2) / image_height
    yolo_width = width / image_width
    yolo_height = height / image_height

    return left, top, width, height, x_center, y_center, yolo_width, yolo_height


def is_in_ignored_region(left, top, width, height, ignored_boxes, threshold=0.5):
    #Check if at least 50% of a bounding box falls inside ignored regions
    box_area = width * height
    if box_area <= 0 or not ignored_boxes:
        return False

    right = left + width
    bottom = top + height
    overlap_area = 0.0

    for ig_left, ig_top, ig_right, ig_bottom in ignored_boxes:
        inter_left = max(left, ig_left)
        inter_top = max(top, ig_top)
        inter_right = min(right, ig_right)
        inter_bottom = min(bottom, ig_bottom)

        if inter_right > inter_left and inter_bottom > inter_top:
            overlap_area += (inter_right - inter_left) * (inter_bottom - inter_top)

    return (overlap_area / box_area) >= threshold
```

```python
#Clear old processed files before rebuilding the dataset
if processed_dir.exists():
    shutil.rmtree(processed_dir)

#Create image and label folders for each split
for split in ['train', 'val', 'test']:
    (processed_dir / 'images' / split).mkdir(parents=True, exist_ok=True)
    (processed_dir / 'labels' / split).mkdir(parents=True, exist_ok=True)

frame_rows = []
annotation_rows = []

#Process one video sequence at a time
for row in sequence_df.itertuples():
    root = ET.parse(row.xml_file).getroot()

    #Read ignored regions to mask out unlabeled vehicles
    ignored_boxes = []
    for box in root.findall('./ignored_region/box'):
        ig_left = float(box.attrib['left'])
        ig_top = float(box.attrib['top'])
        ig_right = ig_left + float(box.attrib['width'])
        ig_bottom = ig_top + float(box.attrib['height'])
        ignored_boxes.append((ig_left, ig_top, ig_right, ig_bottom))

    #Match frame numbers to XML and image files
    xml_frames = {int(frame.attrib['num']): frame for frame in root.findall('./frame')}
    image_files = {
        int(path.stem.replace('img', '')): path
        for path in Path(row.image_folder).glob('img*.jpg')
    }

    for frame_number in sorted(xml_frames.keys()):
        #Use every fifth frame to reduce repeated images
        if (frame_number - 1) % frame_step != 0:
            continue

        if frame_number not in image_files:
            continue

        source_image = image_files[frame_number]
        image = Image.open(source_image).convert('RGB')
        image_width, image_height = image.size

        #Black out ignored regions before saving the image
        if ignored_boxes:
            draw = ImageDraw.Draw(image)
            for ig_left, ig_top, ig_right, ig_bottom in ignored_boxes:
                draw.rectangle(
                    [
                        max(0, ig_left),
                        max(0, ig_top),
                        min(image_width, ig_right),
                        min(image_height, ig_bottom)
                    ],
                    fill=(0, 0, 0)
                )

        file_name = f'{row.sequence}_img{frame_number:05d}'
        output_image = processed_dir / 'images' / row.split / f'{file_name}.jpg'
        output_label = processed_dir / 'labels' / row.split / f'{file_name}.txt'

        image.save(output_image)

        label_lines = []
        frame_object_count = 0

        for target in xml_frames[frame_number].findall('./target_list/target'):
            box = target.find('./box')

            if box is None:
                continue

            converted_box = convert_box(box, image_width, image_height)

            if converted_box is None:
                continue

            left, top, width, height, x_center, y_center, yolo_width, yolo_height = converted_box

            if is_in_ignored_region(left, top, width, height, ignored_boxes):
                continue

            vehicle_type = get_vehicle_type(target)
            class_id = class_ids[vehicle_type]

            label_lines.append(
                f'{class_id} {x_center:.6f} {y_center:.6f} {yolo_width:.6f} {yolo_height:.6f}'
            )

            annotation_rows.append({
                'sequence': row.sequence,
                'frame': frame_number,
                'split': row.split,
                'class_name': vehicle_type,
                'class_id': class_id,
                'left': left,
                'top': top,
                'width': width,
                'height': height,
                'area': (width * height) / (image_width * image_height),
                'image_path': str(output_image)
            })

            frame_object_count += 1

        output_label.write_text('\n'.join(label_lines))

        frame_rows.append({
            'sequence': row.sequence,
            'frame': frame_number,
            'split': row.split,
            'weather': row.weather,
            'image_path': str(output_image),
            'label_path': str(output_label),
            'objects': frame_object_count
        })

frame_df = pd.DataFrame(frame_rows)
annotation_df = pd.DataFrame(annotation_rows)

print('Frames prepared:', len(frame_df))
print('Bounding boxes prepared:', len(annotation_df))
```

7. Save Dataset Files

```python
#Save CSV files for later analysis
frame_df.to_csv(processed_dir / 'frame_manifest.csv', index=False)
annotation_df.to_csv(processed_dir / 'annotations.csv', index=False)
sequence_df[['sequence', 'camera_group', 'weather', 'frames', 'objects', 'split']].to_csv(
    processed_dir / 'sequence_split.csv',
    index=False
)

#Create the YAML file used by YOLO
data_yaml = {
    'path': str(processed_dir.resolve()),
    'train': 'images/train',
    'val': 'images/val',
    'test': 'images/test',
    'names': class_names
}

with open(processed_dir / 'data.yaml', 'w') as file:
    yaml.safe_dump(data_yaml, file, sort_keys=False)

print('Saved:', processed_dir / 'data.yaml')
```

8. Check the Processed Dataset

```python
#Check that every image has a label file
missing_labels = 0
invalid_coordinates = 0

for row in frame_df.itertuples():
    label_path = Path(row.label_path)

    if not label_path.exists():
        missing_labels += 1
        continue

    for line in label_path.read_text().splitlines():
        values = line.split()

        if len(values) != 5:
            invalid_coordinates += 1
            continue

        coordinates = [float(value) for value in values[1:]]

        if not all(0 <= value <= 1 for value in coordinates):
            invalid_coordinates += 1

#Check that a sequence and camera site only appear in one split, weather is populated, and ignored boxes are filtered
sequence_leakage = frame_df.groupby('sequence')['split'].nunique().max()
camera_leakage = sequence_df.groupby('camera_group')['split'].nunique().max()
unknown_weather = int((frame_df['weather'] == 'unknown').sum())
min_test_per_weather = int(sequence_df[sequence_df['split'] == 'test']['weather'].value_counts().min())

ignored_box_leakage = 0
xml_lookup = dict(zip(sequence_df['sequence'], sequence_df['xml_file']))
ignored_lookup = {}
for seq_name, xml_path in xml_lookup.items():
    seq_root = ET.parse(xml_path).getroot()
    ignored_lookup[seq_name] = [
        (
            float(b.attrib['left']),
            float(b.attrib['top']),
            float(b.attrib['left']) + float(b.attrib['width']),
            float(b.attrib['top']) + float(b.attrib['height'])
        )
        for b in seq_root.findall('./ignored_region/box')
    ]

for ann in annotation_df.itertuples():
    if is_in_ignored_region(ann.left, ann.top, ann.width, ann.height, ignored_lookup[ann.sequence]):
        ignored_box_leakage += 1

frame_split_pct = (frame_df['split'].value_counts(normalize=True).reindex(['train', 'val', 'test']) * 100).round(1)

print('Missing labels:', missing_labels)
print('Invalid labels:', invalid_coordinates)
print('Maximum splits per sequence:', sequence_leakage)
print('Maximum splits per camera site:', camera_leakage)
print('Frames with unknown weather:', unknown_weather)
print('Minimum test sequences per weather:', min_test_per_weather)
print('Boxes >=50% inside ignored regions:', ignored_box_leakage)
print('Frame split (%):', frame_split_pct.to_dict())

assert missing_labels == 0
assert invalid_coordinates == 0
assert sequence_leakage == 1
assert camera_leakage == 1
assert unknown_weather == 0
assert min_test_per_weather >= 3
assert ignored_box_leakage == 0

print('Dataset checks passed')
```
